```python
import requests
import pandas as pd
import json

team_id = 147   # Example: New York Yankees

url = f"https://statsapi.mlb.com/api/v1/teams/{team_id}/roster"

response = requests.get(url)
response.raise_for_status()

data = response.json()

# Show top-level keys
print("Top-level JSON keys:")
print(data.keys())
print()

# Extract roster list
roster = data["roster"]

# Convert to dataframe
df = pd.json_normalize(roster)

# Show column names
print("Columns returned from roster endpoint:")
print(df.columns)
print()

# Show first few rows
print("Sample data:")
print(df.head())
```

```python
import json
url_stats = "https://statsapi.mlb.com/api/v1/people/667670/stats?stats=gameLog&group=hitting&season=2025"

response = requests.get(url_stats)
response.raise_for_status()
data = response.json()

print(data.keys())
print(json.dumps(data["stats"], indent=2))                   # step 2
print(data["stats"][0]["splits"][0].keys())
print(json.dumps(data["stats"][0]["splits"][0], indent=2))                             # step 3

```

```python
import requests
import pandas as pd

url_stats = "https://statsapi.mlb.com/api/v1/people/667670/stats?stats=gameLog&group=hitting&season=2025"

response = requests.get(url_stats)
response.raise_for_status()
data = response.json()

print(data.keys())
print(data["stats"])

if not data.get("stats"):
    print("No stats returned for this player / group / season combination.")
else:
    stats_2 = data["stats"][0].get("splits", [])
    df = pd.json_normalize(stats_2)
    print(df.columns)
    print(df.head())
```
