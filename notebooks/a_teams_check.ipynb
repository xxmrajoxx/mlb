```python
import pandas as pd
import requests
import json
```

```python
url_stats_hit = "https://statsapi.mlb.com/api/v1/people/667670/stats?stats=gameLog&group=hitting&season=2025"

response = requests.get(url_stats_hit)
data = response.json()

spilts = data["stats"][0]["splits"]

if not spilts:
    print(f"Player not found")
else:
    print(json.dumps(spilts[0], indent=2))

```

```python

url_stats = "https://statsapi.mlb.com/api/v1/people/667670/stats?stats=gameLog&group=hitting&season=2025"

response = requests.get(url_stats)
response.raise_for_status()

data = response.json()

splits = data["stats"][0]["splits"]

if len(splits) > 5:
    print(json.dumps(splits[5], indent=2))
else:
    print(f"Only {len(splits)} games available")
```

```python
import json 
import pandas 
import requests

url = "https://statsapi.mlb.com/api/v1/teams/147/stats?stats=season&group=hitting&season=2025"
response = requests.get(url)
response.raise_for_status()
data = response.json()

print(json.dumps(data, indent=2))
print(data.keys())

print(data["stats"][0]["splits"][0]["stat"].keys())
```

```python
# to get all columns

all_columns = set()
splits = data["stats"][0]["splits"]

for row in splits:
    stat = row.get("stat", {})
    all_columns.update(stat.keys())
    

print(sorted(all_columns))
```
