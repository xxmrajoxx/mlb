```python
import pandas as pd
import json
import requests

url = "https://statsapi.mlb.com/api/v1/standings?leagueId=103,104&season=2026"
response = requests.get(url)
response.raise_for_status()
data = response.json()

print(data.keys())
print(json.dumps(data, indent=2))

```
